```python
### 1. Import Libraries
import pandas as pd
### 2. Load Dataset
df = pd.read_excel("Dataset for Data Analytics.xlsx")
df.head()
```

```python
### 3. Initial Data Inspection
df.shape
df.info()
```

```python
### 4. Missing Value Analysis
df.isnull().sum()
### 5. Handling Missing Values
df["CouponCode"] = df["CouponCode"].fillna("No Coupon")
```

```python
df.isnull().sum()
```

```python
### 6. Duplicate Check
df.duplicated().sum()
df["OrderID"].duplicated().sum()
```

```python
### 7. Date Validation
df["Date"].min(), df["Date"].max()
```

```python
### 8. Numeric Data Validation
df["Date"].isna().sum()
```

```python
pd.to_datetime(df["Date"], errors="coerce").isna().sum()
```

```python
df.describe()
```

```python
(df["Quantity"] * df["UnitPrice"] - df["TotalPrice"]).abs().max()
```

```python
### 9. Text Data Validation
text_columns = [
    "Product",
    "ShippingAddress",
    "PaymentMethod",
    "OrderStatus",
    "CouponCode",
    "ReferralSource"
]

for col in text_columns:
    print("\n", col)
    print(df[col].unique())
```

```python
df.info()
```

```python
### 10. Final Data Validation
print("FINAL DATA VALIDATION")
print("---------------------")

print("Total rows:", len(df))
print("Total columns:", len(df.columns))
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate OrderIDs:", df["OrderID"].duplicated().sum())
print("Invalid dates:", pd.to_datetime(df["Date"], errors="coerce").isna().sum())
```

```python
### 11. Export Cleaned Dataset### 11. Export Cleaned Dataset
df.to_excel("cleaned_dataset.xlsx", index=False)
```
